```python
import pandas as pd

# Create a sample dataset
data = {
    'Name': ['Amal', 'Kamal', 'Nimal', 'Saman', 'Sunil'],
    'Age': [21, 22, 20, 23, 21],
    'Marks': [75, 85, 65, 90, 80]
}

df = pd.DataFrame(data)

# Display dataset
print(df)

# Calculate average marks
print("Average Marks:", df['Marks'].mean())

# Find highest marks
print("Highest Marks:", df['Marks'].max())
```
